```python
import pandas as pd
f = pd.read_parquet("../data/processed/features.parquet")
e = pd.read_parquet("../data/processed/entries.parquet")
e = e[e["round"] == "final"][["entry_id", "artist_age", "artist_sex"]]
full = f.merge(e, on="entry_id", how="left", validate="m:1")

# full["artist_sex"].unique
full.to_csv("features_full.csv", index=False)
```
